```python
import os
from zipfile import ZipFile

from joblib import Parallel, cpu_count, delayed
from tqdm.notebook import tqdm

DATASET = "./GarminRawData/"
cpu_count()
```

# Convert .fit to .json

```python
files = []
for dirName in os.listdir(DATASET):
    basedir = os.path.join(DATASET, dirName)
    if basedir[-4:] == ".zip":
        print(f"""Extracting: {basedir}""")
        with ZipFile(basedir, "r") as zipObj:
            zipObj.extractall(DATASET)
        basedir = basedir[:-4]
    if os.path.isdir(basedir):
        for fName in tqdm(os.listdir(basedir)):
            if fName[-4:] == ".fit":
                fName = os.path.join(basedir, fName)
                print(f"""Converting file: {fName}""")
                os.system(f"""fitjson --pretty -o {fName[:-4]}.json {fName}""")
```

```python
files = []


def convertFit(dirName):
    print(dirName)
    basedir = os.path.join(DATASET, dirName)
    if basedir[-4:] == ".zip":
        print(f"""Extracting: {basedir}""")
        with ZipFile(basedir, "r") as zipObj:
            zipObj.extractall(DATASET)
        basedir = basedir[:-4]

    if os.path.isdir(basedir):
        for fName in tqdm(os.listdir(basedir)):
            if fName[-4:] == ".fit":
                fName = os.path.join(basedir, fName)
                print(f"""Converting file: {fName}""")
                os.system(f"""fitjson --pretty -o {fName[:-4]}.json {fName}""")


Parallel(n_jobs=cpu_count() - 1)(
    delayed(convertFit)(dirName) for dirName in os.listdir(DATASET)
)
```
